```python
class Vocabulary:

    def __init__(self):
        self.word2idx = {
            '<pad>': 0,
            '<sos>': 1,
            '<eos>': 2,
            '<unk>': 3
        }
        self.idx2word = {
            0: '<pad>',
            1: '<sos>',
            2: '<eos>',
            3: '<unk>'
        }

    def build(self, sentences):
        freq = {}
        for sentence in sentences:
            for word in sentence.split():
                if word not in freq:
                    freq[word] = 1
                else:
                    freq[word] += 1

        for word, count in freq.items():
            if count >= 2:
                idx = len(self.word2idx)
                self.word2idx[word] = idx
                self.idx2word[idx]  = word

    def encode(self, sentence):
        words = sentence.split()
        return [self.word2idx.get(word, 3)
                for word in words]

    def decode(self, indices):
        words = []
        for idx in indices:
            word = self.idx2word.get(idx, '<unk>')
            if word == '<eos>':
                break
            if word not in ['<sos>', '<pad>']:
                words.append(word)
        return ' '.join(words)

    def __len__(self):
        return len(self.word2idx)


# English Vocabulary
eng_vocab = Vocabulary()
eng_vocab.build(english_sentences)
print(f"English vocabulary: {len(eng_vocab)}")

# Hindi Vocabulary
hin_vocab = Vocabulary()
hin_vocab.build(hindi_sentences)
print(f"Hindi vocabulary: {len(hin_vocab)}")
```

```python
# Pehle TranslationDataset banao
from torch.utils.data import Dataset, DataLoader
import torch

class TranslationDataset(Dataset):
    def __init__(self, eng_sentences, hin_sentences,
                 eng_vocab, hin_vocab):
        self.eng_sentences = eng_sentences
        self.hin_sentences = hin_sentences
        self.eng_vocab     = eng_vocab
        self.hin_vocab     = hin_vocab

    def __len__(self):
        return len(self.eng_sentences)

    def __getitem__(self, idx):
        eng = self.eng_sentences[idx]
        hin = self.hin_sentences[idx]
        eng_ids = [1] + self.eng_vocab.encode(eng) + [2]
        hin_ids = [1] + self.hin_vocab.encode(hin) + [2]
        return eng_ids, hin_ids


def collate_fn(batch):
    src_batch, tgt_batch = zip(*batch)
    src_max = max(len(s) for s in src_batch)
    tgt_max = max(len(t) for t in tgt_batch)
    src_padded = [list(s) + [0] * (src_max - len(s)) for s in src_batch]
    tgt_padded = [list(t) + [0] * (tgt_max - len(t)) for t in tgt_batch]
    return torch.LongTensor(src_padded), torch.LongTensor(tgt_padded)


# Pehle TranslationDataset banao
translation_dataset = TranslationDataset(
    english_sentences,
    hindi_sentences,
    eng_vocab,
    hin_vocab
)

# Ab DataLoader mein do!
train_loader = DataLoader(
    translation_dataset,  # ← Ye sahi hai!
    batch_size=64,
    shuffle=True,
    collate_fn=collate_fn
)

print(f"Total batches: {len(train_loader)}")
```

```python
import torch
import torch.nn as nn

class RNNEncoder(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()

        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.rnn       = nn.RNN(embed_dim, hidden_dim,
                                batch_first=True)

    def forward(self, src):
        embedded        = self.embedding(src)
        outputs, hidden = self.rnn(embedded)
        return hidden
```

```python
class RNNDecoder (nn.Module):
  def __init__ (self , vocab_size , embed_dim , hidden_dim):
    super().__init__()

    self.embedding = nn.Embedding(vocab_size , embed_dim)
    self.rnn = nn.RNN(embed_dim , hidden_dim , batch_first=True)
    self.fc = nn.Linear(hidden_dim , vocab_size)


  def forward(self , tgt , hidden):
    embedded        = self.embedding(tgt)
    output , hidden = self.rnn(embedded , hidden)
    prediction = self.fc(output)
    return prediction , hidden
#
```

```python
class RNNSeq2Seq(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, tgt):
        hidden = self.encoder(src)
        output, _ = self.decoder(tgt, hidden)
        return output
```

```python
# Model banao
encoder = RNNEncoder(
    vocab_size = len(eng_vocab),
    embed_dim  = 256,
    hidden_dim = 512
)

decoder = RNNDecoder(
    vocab_size = len(hin_vocab),
    embed_dim  = 256,
    hidden_dim = 512
)

model = RNNSeq2Seq(encoder, decoder)

# Parameters count karo
total = sum(p.numel() for p in model.parameters())
print(f"Total parameters: {total:,}")
```

```python
import torch.optim as optim

# Device set karo
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

# Model, Loss, Optimizer
model     = RNNSeq2Seq(encoder, decoder).to(device)
criterion = nn.CrossEntropyLoss(ignore_index=0)
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("✅ Ready for training!")
```

```python
def train_epoch(model, loader, optimizer, criterion, device):
    model.train()
    total_loss = 0

    for src, tgt in loader:
        src = src.to(device)
        tgt = tgt.to(device)

        optimizer.zero_grad()

        output = model(src, tgt)

        output = output[:, :-1].reshape(-1, len(hin_vocab))
        target = tgt[:, 1:].reshape(-1)

        loss = criterion(output, target)
        loss.backward()

        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(loader)
```

```python
EPOCHS = 10
best_loss = float('inf')

for epoch in range(1, EPOCHS + 1):
    train_loss = train_epoch(
        model, train_loader,
        optimizer, criterion, device
    )

    print(f"Epoch {epoch}/{EPOCHS} | Loss: {train_loss:.4f}")

    if train_loss < best_loss:
        best_loss = train_loss
        torch.save(model.state_dict(),
                   DRIVE_PATH + 'best_rnn.pt')
        print(f"✅ Best model saved!")
```

```python
def translate_rnn(model, sentence, eng_vocab,
                  hin_vocab, device, max_len=50):
    model.eval()

    tokens = [1] + eng_vocab.encode(sentence) + [2]
    src    = torch.LongTensor(tokens).unsqueeze(0).to(device)

    hidden = model.encoder(src)

    tgt_token = torch.LongTensor([[1]]).to(device)
    result    = []

    for _ in range(max_len):
        output, hidden = model.decoder(tgt_token, hidden)
        pred_token     = output.argmax(2)[:, -1]

        if pred_token.item() == 2:
            break

        result.append(pred_token.item())
        tgt_token = pred_token.unsqueeze(0)

    return hin_vocab.decode(result)


# Test karo
test_sentences = [
    "my name is akshay"
]

for sentence in test_sentences:
    translation = translate_rnn(
        model, sentence,
        eng_vocab, hin_vocab, device
    )
    print(f"English : {sentence}")
    print(f"Hindi   : {translation}")
    print()
```

```python
class LSTMEncoder (nn.Module):
  def __init__(self, vocab_size ,  embed_dim , hidden_dim):
    super().__init__()
    self.embedding = nn.Embedding (vocab_size , embed_dim)
    self.rnn = nn.LSTM(embed_dim , hidden_dim , batch_first= True)

  def forward (self ,src):
    embedded = self.embedding (src)
    outputs , (hidden, cell ) = self.rnn(embedded)
    return hidden , cell
```

```python
class LSTMDecoder(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.rnn = nn.LSTM(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, tgt, hidden,cell):
        embedded = self.embedding(tgt)
        output, (hidden,cell) = self.rnn(embedded, (hidden, cell))
        prediction = self.fc(output)
        return prediction, hidden, cell
```

```python
class LSTMseq2seq (nn.Module):
  def __init__(self , encoder , decoder):
    super().__init__()
    self.encoder = encoder
    self.decoder = decoder

  def forward (self ,src ,tgt):
    hidden , cell = self.encoder(src)
    output,_,_= self.decoder(tgt , hidden, cell)

    return output

```

```python
class GRUEncoder (nn.Module):
  def __init__ (self , vocab_size , embed_dim , hidden_dim):
    super(). __init__ ()
    self.embedding = nn.Embedding (vocab_size , embed_dim)
    self.rnn = nn.GRU(embed_dim , hidden_dim ,batch_first =True)
  def forward (self ,src ):
    embedded = self.embedding (src)
    output , hidden = self.rnn (embedded)
    return hidden

```

```python
class GRUDecoder(nn.Module):
    def __init__(self, vocab_size, embed_dim, hidden_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.rnn = nn.GRU(embed_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, vocab_size)

    def forward(self, tgt, hidden):
        embedded = self.embedding(tgt)
        output, hidden = self.rnn(embedded, hidden)
        prediction = self.fc(output)
        return prediction, hidden
```

```python
class GRUSeq2Seq(nn.Module):
    def __init__(self, encoder, decoder):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder

    def forward(self, src, tgt):
        hidden = self.encoder(src)
        output, _ = self.decoder(tgt, hidden)
        return output
```

```python
class TransformerModel (nn.Module):
  def __init__ (self , src_vocab_size ,tgt_vocab_size , embed_dim ,nhead,num_layers, dropout):
    super().__init__()
    self.src_embedding = nn.Embedding (src_vocab_size , embed_dim)
    self.tgt_embedding = nn.Embedding (tgt_vocab_size , embed_dim)
    self.transformer = nn.Transformer (d_model=embed_dim,nhead=nhead,num_encoder_layers=num_layers, num_decoder_layers= num_layers,dropout=dropout,batch_first=True)
    self.fc = nn.Linear(embed_dim , tgt_vocab_size)

  def forward(self, src, tgt):
    src_emb = self.src_embedding(src)  # English embedding
    tgt_emb = self.tgt_embedding(tgt)  # Hindi embedding
    output = self.transformer(src_emb, tgt_emb)  # Transformer
    prediction = self.fc(output)       # Final output
    return prediction
```

```python
def translate_transformer(model, sentence, eng_vocab, hin_vocab, device, max_len=50):
    model.eval()
    with torch.no_grad():
        tokens = [1] + eng_vocab.encode(sentence) + [2]
        src = torch.LongTensor(tokens).unsqueeze(0).to(device)
        tgt = torch.LongTensor([[1]]).to(device)

        for _ in range(max_len):
            output = model(src, tgt)
            pred_token = output.argmax(2)[:, -1]
            if pred_token.item() == 2: break
            tgt = torch.cat([tgt, pred_token.unsqueeze(0)], dim=1)

        return hin_vocab.decode(tgt.squeeze(0).tolist())
```

```python
def translate_rnn_gru(model, sentence, eng_vocab, hin_vocab, device, max_len=50):
    model.eval()
    with torch.no_grad():
        tokens = [1] + eng_vocab.encode(sentence) + [2]
        src = torch.LongTensor(tokens).unsqueeze(0).to(device)
        hidden = model.encoder(src)
        tgt_token = torch.LongTensor([[1]]).to(device)
        result = []
        for _ in range(max_len):
            output, hidden = model.decoder(tgt_token, hidden)
            pred_token = output.argmax(2)[:, -1]
            if pred_token.item() == 2: break
            result.append(pred_token.item())
            tgt_token = pred_token.unsqueeze(0)
        return hin_vocab.decode(result)


def translate_lstm(model, sentence, eng_vocab, hin_vocab, device, max_len=50):
    model.eval()
    with torch.no_grad():
        tokens = [1] + eng_vocab.encode(sentence) + [2]
        src = torch.LongTensor(tokens).unsqueeze(0).to(device)
        hidden, cell = model.encoder(src)
        tgt_token = torch.LongTensor([[1]]).to(device)
        result = []
        for _ in range(max_len):
            output, hidden, cell = model.decoder(tgt_token, hidden, cell)
            pred_token = output.argmax(2)[:, -1]
            if pred_token.item() == 2: break
            result.append(pred_token.item())
            tgt_token = pred_token.unsqueeze(0)
        return hin_vocab.decode(result)
```

```python
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = MarianTokenizer.from_pretrained(model_name)
pretrained_model = MarianMTModel.from_pretrained(model_name)

def translate_pretrained(sentence):
    inputs = tokenizer(sentence, return_tensors="pt", padding=True)
    translated = pretrained_model.generate(**inputs)
    return tokenizer.decode(translated[0], skip_special_tokens=True)
```
